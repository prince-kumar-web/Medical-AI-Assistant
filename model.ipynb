```python
import pandas as pd

df = pd.read_csv("data/Training.csv")

# Target alag
X = df.drop("prognosis", axis=1)
y = df["prognosis"]

print(X.shape)
print(y.shape)
```

```python
import os

print(os.getcwd())
print(os.listdir())
```

```python
X = df.drop("prognosis", axis=1)
y = df["prognosis"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
def symptoms_to_text(row):

    active_symptoms = []

    for symptom in X.columns:

        if row[symptom] == 1:
            active_symptoms.append(
                symptom.replace("_", " ")
            )

    return " ".join(active_symptoms)


df["symptom_text"] = X.apply(
    symptoms_to_text,
    axis=1
)
```

```python
print(
    df[["symptom_text", "prognosis"]].head()
)
```

```python
import transformers

print(transformers.__version__)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

print("BERT tokenizer loaded successfully")
```

```python
sample_text = df["symptom_text"].iloc[0]

print("TEXT:")
print(sample_text)

tokens = tokenizer(
    sample_text,
    padding="max_length",
    truncation=True,
    max_length=64,
    return_tensors="pt"
)

print("\nINPUT IDS SHAPE:")
print(tokens["input_ids"].shape)

print("\nATTENTION MASK SHAPE:")
print(tokens["attention_mask"].shape)
```

```python
import torch
from transformers import AutoModel

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

bert_model = AutoModel.from_pretrained(
    "bert-base-uncased"
)

bert_model = bert_model.to(device)
bert_model.eval()

print("BERT model loaded successfully")
```

```python
sample_text = df["symptom_text"].iloc[0]

inputs = tokenizer(
    sample_text,
    padding=True,
    truncation=True,
    max_length=64,
    return_tensors="pt"
)

inputs = {
    key: value.to(device)
    for key, value in inputs.items()
}

with torch.no_grad():

    outputs = bert_model(**inputs)

    embedding = outputs.last_hidden_state[:, 0, :]

print("Text:", sample_text)
print("Embedding shape:", embedding.shape)
```

```python
outputs.last_hidden_state[:, 0, :]
```

```python
import numpy as np
from torch.utils.data import DataLoader, TensorDataset

# Text data
texts = df["symptom_text"].tolist()

# Batch size
batch_size = 16

all_embeddings = []

for i in range(0, len(texts), batch_size):

    batch_texts = texts[i:i + batch_size]

    inputs = tokenizer(
        batch_texts,
        padding=True,
        truncation=True,
        max_length=64,
        return_tensors="pt"
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = bert_model(**inputs)

        # CLS token embedding
        batch_embeddings = (
            outputs.last_hidden_state[:, 0, :]
        )

    all_embeddings.append(
        batch_embeddings.cpu().numpy()
    )

    if (i // batch_size) % 10 == 0:
        print(f"Processed: {min(i + batch_size, len(texts))}/{len(texts)}")


# Combine all batches
X_bert = np.vstack(all_embeddings)

print("Final BERT embedding shape:", X_bert.shape)
```

```python
y = df["prognosis"]

print("X:", X_bert.shape)
print("y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_bert,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("Random Forest trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_precision = precision_score(
    y_test,
    rf_pred,
    average="weighted"
)

rf_recall = recall_score(
    y_test,
    rf_pred,
    average="weighted"
)

rf_f1 = f1_score(
    y_test,
    rf_pred,
    average="weighted"
)

print("Random Forest Results")
print("---------------------")
print("Accuracy :", round(rf_accuracy, 4))
print("Precision:", round(rf_precision, 4))
print("Recall   :", round(rf_recall, 4))
print("F1 Score :", round(rf_f1, 4))
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        rf_pred
    )
)
```

```python
def predict_disease_bert(text):

    inputs = tokenizer(
        text,
        padding=True,
        truncation=True,
        max_length=64,
        return_tensors="pt"
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        outputs = bert_model(**inputs)

    embedding = outputs.last_hidden_state[:, 0, :]
    embedding = embedding.cpu().numpy()

    prediction = bert_lr_model.predict(embedding)

    return prediction[0]
```

```python
def predict_with_confidence(text):

    inputs = tokenizer(
        text,
        padding=True,
        truncation=True,
        max_length=64,
        return_tensors="pt"
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        outputs = bert_model(**inputs)

    embedding = outputs.last_hidden_state[:, 0, :]
    embedding = embedding.cpu().numpy()

    probabilities = bert_lr_model.predict_proba(embedding)[0]

    classes = bert_lr_model.classes_

    top_indices = probabilities.argsort()[-3:][::-1]

    results = []

    for index in top_indices:
        results.append(
            (
                classes[index],
                round(probabilities[index] * 100, 2)
            )
        )

    return results
```
